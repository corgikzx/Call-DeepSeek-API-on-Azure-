# 使用部署在Azure上的DeepSeek API服务

## 前置要求


* 创建属于你的 Azure subscription，并在 Azure AI Foundry 上部署 DeepSeek-R1 模型，请参考 https://learn.microsoft.com/en-us/azure/ai-studio/how-to/deploy-models-deepseek?pivots=programming-language-python
* 获取 endpoint 和 API key
* 在本地安装 Python 并安装所需要的各种包（主要是 azure-ai-inference）

## 代码

### 设置模型部分

请依次运行以下两个 cell 中的代码，其中第二个 cell 请注意修改填入你自己的 endpoint 和 key

```python
def ask_question(question):
    global messages
    # 添加用户问题到对话上下文
    messages.append(UserMessage(content=question))
    
    # 获取模型回复
    response = model.complete(
        messages=messages,
        model="deepseek-r1",  # 指定模型名称
        model_extras={
            "safe_mode": True
        }
        temperature=0,
        top_p=1,
        stream=True,
    )
    
    # 提取模型回复
    answer = response.choices[0].message.content
    
    # 添加模型回复到对话上下文
    messages.append({"role": "assistant", "content": answer})
    
    return answer
```

```python
import os
from azure.ai.inference import ChatCompletionsClient
from azure.core.credentials import AzureKeyCredential

# 设置环境变量（仅用于示例，实际使用时应在操作系统中设置）
os.environ["AZURE_OPENAI_ENDPOINT"] = "your Azure endpoint, other take the form as https://<resource-name>.services.ai.azure.com/models"
os.environ["AZURE_OPENAI_API_KEY"] = "you Azure AI API key"

# 从环境变量中获取值
endpoint = os.environ["AZURE_OPENAI_ENDPOINT"]
api_key = os.environ["AZURE_OPENAI_API_KEY"]

model = ChatCompletionsClient(
    endpoint=endpoint,
    credential=AzureKeyCredential(api_key),
)

from azure.ai.inference.models import SystemMessage, UserMessage

```

### 初始化部分

以下 cell 为初始化代码, 第一次运行时请先运行之，后续如无需 deepseek 在回答问题时回顾你之前的问题和回答，也请运行该代码

```python
# 初始化对话上下文
messages = [
    SystemMessage(content="You are a helpful assistant.")
]
```

### 聊天部分

请在以下代码的 question 字符串填入你的问题并运行该 cell

```python
# 示例提问
question1 = "How many languages are in the world?"
answer1 = ask_question(question1)
print(f"Q: {question1}\nA: {answer1}\n")
```

如需追加提问，请复制以上代码到一个新的 cell 中，修改问题并运行，这样deepseek 将会在回答问题时回顾你之前的问题和回答

```python
question2 = "Which language is the most widely spoken?"
answer2 = ask_question(question2)
print(f"Q: {question2}\nA: {answer2}\n")
```
